**(c)** Compute a geometric formula:

<font size = "3">

The **frustum** of a cone is the shape formed when the top of a cone is cut off parallel to the base, leaving two circular faces of different radii. (Think of a lampshade or a bucket)

The formula for the volume of a frustum of a cone is:


$$V = \frac{1}{3}\pi h (r_1^2 + r_1 r_2 + r_2^2)$$


Where:

- $V$ is the volume of the cylinder
- $r_1$ is the radius of the bottom base
- $r_2$ is the radius of the top base
- $h$ is the height of the frustrum


Calculate $V$ when:

- $r_1 = 6$
- $r_2 = 3$
- $h = 10$


To get full points:
- Define Python variables corresponding to the values of $r_1$, $r_2$, and $h$ above.

- Use your variables to calculate the volume ($V$) using the formula above. This value should also be assigned to a variable.

- Print out the value of the computed volume to the screen.

**Note:** $\pi \neq 3.14$

```python
# your code here
import numpy as np
r_1 = 6 
r_2 = 3 
h = 10 
 

V = (1/3) * np.pi * h *(r_1**2 +r_1*r_2 +r_2**2)
print(V)

```

**(e)** Using NumPy arrays, create a **line** plot of the function

$$f(x) = \frac{1}{\sqrt{2\pi}}\mathrm{e}^{-x^2}$$

at a set of 75 evenly spaced points in the interval $-3 \leq x \leq 3$ using `numpy.linspace()`.

Label the axes and give it a title (this is the probability density function (pdf) of the standad normal probability distribution). Don't forget to suppress extraneous output with `plt.show()`

```python
# your code here
import matplotlib.pyplot as plt
x_vals = np.linspace(-3, 3, 75)
y_vals = (1/np.sqrt(2*np.pi))*(np.e**(-x_vals**2))
plt.plot(x_vals,y_vals)
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.title("Standard Normal Probability Distribution")
plt.show()
```
