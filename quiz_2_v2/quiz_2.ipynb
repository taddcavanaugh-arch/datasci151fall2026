**Question (B)** Defining a variable through a piecewise-defined expression

<font size = "3">

Let $I$ denote the income of a single person, measured in thousands of dollars (so if $I = 100$, the person made $100,000). Let $T$ denote the amount of income tax that person paid, again in thousands of dollars. The amount of tax is determined by the following progressive tax bracket:


$$
T = 
\begin{cases}
0.1 \times I & \text{if }  0 \leq I \leq 12.5\\[6pt]
1.25 + 0.12 \times (I - 12.5) & \text{if } 12.5 < I \leq 48.5 \\[6pt]
5.57 + 0.22 \times (I - 48.5) & \text{if } 48.5 < I \leq 103.5 \\[6pt]
17.67 + 0.25 \times (I - 103.5) & \text{if } 103.5 < I \leq 197.5 \\[6pt]
41.17 + 0.32 \times (I - 197.5) & \text{if } 197.5 < I
\end{cases}
$$

In the code cell below, a variable called `income_in_thousands` is randomly generated between $0$ and $400$. Using `if`, `elif`, and/or `else` statements, write code that defines a variable `tax_paid` that is determined by the expression above.

````python
import random

income_in_thousands = round(random.uniform(0, 400), 3)
print(income_in_thousands)

tax_paid = 0

if 0<= income_in_thousands <= 12.5:
    tax_paid = tax_paid + 0.1 * income_in_thousands
elif 12.5< income_in_thousands <= 48.5:
    tax_paid = tax_paid + 1.25 + 0.12(income_in_thousands-12.5)
elif 48.5 < income_in_thousands <= 103.5:
    tax_paid = tax_paid + 5.57 + 0.22(income_in_thousands-48.5)
elif 103.5 < income_in_thousands <= 197.5:
    tax_paid = tax_paid + 17.67 + 0.25(income_in_thousands-103.5)
elif income_in_thousands > 197.5:
    tax_paid = tax_paid + 41.17 + 0.32(income_in_thousands-197.5)
# your code here
print(income_in_thousands)
print(tax_paid)

````

- Tests if the number satisfies $10 < n < 15$

````python
import random

lo = 1
hi = 20

random_int = random.randint(lo, hi)
print(random_int)

# Your code here
if (1 <= random_int) and (random_int < 5):
    print("In the interval [1, 5)") 
elif (5 <= random_int) and (random_int <= 10):
    print("In the interval [5, 10]")
elif (10 < random_int) and (random_int < 15):
    print("In the interval (10, 15)")
else:
    print("Greater than or equal to 15")

````
